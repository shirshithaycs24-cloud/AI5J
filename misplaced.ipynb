```python
import heapq

# Initial state
initial = (2, 8, 3,
           1, 6, 4,
           7, 0, 5)

# Goal state
goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


# Display puzzle
def display(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


# h(n) = Number of misplaced tiles
def misplaced(state):
    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1

    return count


# Generate possible next states
def neighbors(state):

    result = []

    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    moves = [
        (-1, 0),  # Up
        (1, 0),   # Down
        (0, -1),  # Left
        (0, 1)    # Right
    ]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            new_state = list(state)

            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            result.append(tuple(new_state))

    return result


# A* Search
def astar():

    # (f, g, state)
    open_list = []

    g = 0
    h = misplaced(initial)
    f = g + h

    heapq.heappush(open_list, (f, g, initial))

    parent = {initial: None}
    cost = {initial: 0}

    while open_list:

        f, g, current = heapq.heappop(open_list)

        print("Current Node:")
        display(current)

        print("g =", g)
        print("h =", misplaced(current))
        print("f =", f)
        print("----------------")

        # Goal test
        if current == goal:

            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()

            print("\nSOLUTION PATH")
            print("=============")

            for i, state in enumerate(path):
                print("Step", i)
                display(state)

            print("Total cost =", len(path) - 1)

            return

        # Generate successors
        for next_state in neighbors(current):

            new_g = g + 1

            if next_state not in cost or new_g < cost[next_state]:

                cost[next_state] = new_g

                h = misplaced(next_state)
                f = new_g + h

                parent[next_state] = current

                heapq.heappush(
                    open_list,
                    (f, new_g, next_state)
                )

    print("No solution")


# Run program
astar()
```
