```python

board = [' ' for _ in range(9)]


def print_board():
    print()
    print(f" {board[0]} | {board[1]} | {board[2]}")
    print("---+---+---")
    print(f" {board[3]} | {board[4]} | {board[5]}")
    print("---+---+---")
    print(f" {board[6]} | {board[7]} | {board[8]}")
    print()

def check_win(player):
    winning_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_positions:
        if board[a] == player and board[b] == player and board[c] == player:
            return True

    return False


def is_draw():
    return ' ' not in board


def minimax(is_maximizing):

    if check_win('O'):
        return 1

    if check_win('X'):
        return -1

    if is_draw():
        return 0

    if is_maximizing:
        best_score = -float('inf')

        for i in range(9):
            if board[i] == ' ':
                board[i] = 'O'

                score = minimax(False)

                board[i] = ' '

                best_score = max(best_score, score)

        return best_score

    else:
        best_score = float('inf')

        for i in range(9):
            if board[i] == ' ':
                board[i] = 'X'

                score = minimax(True)

                board[i] = ' '

                best_score = min(best_score, score)

        return best_score


def computer_move():
    best_score = -float('inf')
    best_move = None

    for i in range(9):
        if board[i] == ' ':

            board[i] = 'O'

            score = minimax(False)

            board[i] = ' '

            if score > best_score:
                best_score = score
                best_move = i

    board[best_move] = 'O'


def human_move():
    while True:
        try:
            position = int(input("Enter your move (1-9): "))

            if position < 1 or position > 9:
                print("Please enter a number between 1 and 9.")
                continue

            index = position - 1

            if board[index] != ' ':
                print("That position is already occupied.")
                continue

            board[index] = 'X'
            break

        except ValueError:
            print("Please enter a valid number.")

def play_game():

    print("===== TIC-TAC-TOE =====")
    print("Human    : X")
    print("Computer : O")

    print("\nBoard positions:")
    print(" 1 | 2 | 3")
    print("---+---+---")
    print(" 4 | 5 | 6")
    print("---+---+---")
    print(" 7 | 8 | 9")

    while True:

        human_move()
        print_board()

        if check_win('X'):
            print("Human wins!")
            break

        if is_draw():
            print("Game is a draw!")
            break

        print("Computer is thinking...")
        computer_move()
        print_board()

        if check_win('O'):
            print("Computer wins!")
            break

        if is_draw():
            print("Game is a draw!")
            break


play_game()
```

```python

room_A = input("Enter status of Room A (Clean/Dirty): ").capitalize()
room_B = input("Enter status of Room B (Clean/Dirty): ").capitalize()

vacuum = input("Enter vacuum position (A/B): ").upper()

print("\nInitial State:")
print("Room A:", room_A)
print("Room B:", room_B)
print("Vacuum Position:", vacuum)


def vacuum_agent(room_A, room_B, vacuum):

    print("\n--- Vacuum Cleaner Actions ---")

    if vacuum == 'A':

        if room_A == "Dirty":
            print("Vacuum is in Room A.")
            print("Action: SUCK")
            room_A = "Clean"

        if room_B == "Dirty":
            print("Action: MOVE RIGHT")
            vacuum = 'B'

            print("Vacuum is now in Room B.")
            print("Action: SUCK")
            room_B = "Clean"

    elif vacuum == 'B':

        if room_B == "Dirty":
            print("Vacuum is in Room B.")
            print("Action: SUCK")
            room_B = "Clean"

        if room_A == "Dirty":
            print("Action: MOVE LEFT")
            vacuum = 'A'

            print("Vacuum is now in Room A.")
            print("Action: SUCK")
            room_A = "Clean"

    return room_A, room_B, vacuum


room_A, room_B, vacuum = vacuum_agent(
    room_A, room_B, vacuum
)


print("\nFinal State:")
print("Room A:", room_A)
print("Room B:", room_B)
print("Vacuum Position:", vacuum)

if room_A == "Clean" and room_B == "Clean":
    print("\nGoal achieved!")
    print("Both rooms are clean.")
else:
    print("\nGoal not achieved.")
```
