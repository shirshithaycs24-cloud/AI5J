```python

initial = (2, 8, 3,
           1, 6, 4,
           7, 0, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


def display(state):
    for i in range(0, 9, 3):
        print(state[i], state[i + 1], state[i + 2])
    print()


def get_successors(state):

    successors = []

    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    if row > 0:
        new_state = list(state)
        new_blank = blank - 3
        new_state[blank], new_state[new_blank] = \
            new_state[new_blank], new_state[blank]
        successors.append(tuple(new_state))

    if row < 2:
        new_state = list(state)
        new_blank = blank + 3
        new_state[blank], new_state[new_blank] = \
            new_state[new_blank], new_state[blank]
        successors.append(tuple(new_state))

    if col > 0:
        new_state = list(state)
        new_blank = blank - 1
        new_state[blank], new_state[new_blank] = \
            new_state[new_blank], new_state[blank]
        successors.append(tuple(new_state))

    if col < 2:
        new_state = list(state)
        new_blank = blank + 1
        new_state[blank], new_state[new_blank] = \
            new_state[new_blank], new_state[blank]
        successors.append(tuple(new_state))

    return successors


def depth_limited_search(node, goal, depth_limit, current_depth, path):

    if node == goal:
        return path

    if current_depth == depth_limit:
        return None

    for child in get_successors(node):

        if child not in path:

            result = depth_limited_search(
                child,
                goal,
                depth_limit,
                current_depth + 1,
                path + [child]
            )

            if result is not None:
                return result

    return None


def iterative_deepening_search(initial, goal):

    depth_limit = 0

    while True:

        print("Searching at depth:", depth_limit)

        result = depth_limited_search(
            initial,
            goal,
            depth_limit,
            0,
            [initial]
        )

        if result is not None:
            return result

        depth_limit += 1


solution = iterative_deepening_search(initial, goal)


print("\nSOLUTION FOUND")
print("================")

for i, state in enumerate(solution):

    print("Step", i)
    display(state)

print("Total moves:", len(solution) - 1)
```
